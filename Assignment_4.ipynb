```python
a=[[1,2],             #using list
   [3,4]]
b=[[5,6],
   [7,8]]
result=[[0,0],
        [0,0]]
for i in range(2):
    for j in range(2):
        result[i][j]=a[i][j]+b[i][j]
print("Sum of two matrices")
for row in result:
    print(row)
```

```python
import numpy as np    #using numpy
a=np.array([[1,2],
            [3,4]])
b=np.array([[5,6],
            [7,8]])
result=a+b
print("sum of two matrices=",result)
```

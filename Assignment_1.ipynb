```python
print("==List Operations==")
list1=[10,20,30,40,20,51]
a=list1+[60]
b=list1.count(20)
c=list1.pop()
d=list1*2
e=list1.index(40)

print(a)
print(b)
print(c)
print(d)
print(e)
```

```python
print("==Tuple Operations==")
tup1=(1,2,3,4,5,2)
print(tup1.index(4))
print(tup1.count(2))
print(len(tup1))
print(tup1*2)
```

```python
print("==Dictionary Operations==")
dict1={"Name":"Dyon","age":"18","College":"MIT"}
print(len(dict1)) #prints length of dict1
dict1["Name"]="Razeen" #updates name
dict1['Branch']="CSE-AIML" #adds value
print(dict1.keys())
print(dict1.values())
```
